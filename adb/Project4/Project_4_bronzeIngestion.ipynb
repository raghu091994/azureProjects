```python
import requests

files = {
    "2019.csv": "https://raw.githubusercontent.com/MicrosoftLearning/mslearn-databricks/main/data/2019.csv",
    "2020.csv": "https://raw.githubusercontent.com/MicrosoftLearning/mslearn-databricks/main/data/2020.csv",
    "2021.csv": "https://raw.githubusercontent.com/MicrosoftLearning/mslearn-databricks/main/data/2021.csv"
}

base_path = "/Volumes/adbrag/project4_schema/project4_volume/raw/"

for file_name, url in files.items():
    response = requests.get(url)
    response.raise_for_status() #no file warning

    with open(base_path + file_name, "wb") as f: #writing binary
        f.write(response.content)

    print(f"Downloaded {file_name}")
```
