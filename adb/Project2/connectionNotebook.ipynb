```python
dbutils.secrets.get("adls-scope", "sp-databricksapp-client-id")
dbutils.secrets.get("adls-scope", "sp-databricksapp-clientsecret")
dbutils.secrets.get("adls-scope", "sp-databricksapp-tenantid")
```

```python
storage_account = "adlsg2rag"

client_id = dbutils.secrets.get("adls-scope", "sp-databricksapp-client-id")
client_secret = dbutils.secrets.get("adls-scope", "sp-databricksapp-clientsecret")
tenant_id = dbutils.secrets.get("adls-scope", "sp-databricksapp-tenantid")

spark.conf.set(
    f"fs.azure.account.auth.type.{storage_account}.dfs.core.windows.net",
    "OAuth"
)
spark.conf.set(
    f"fs.azure.account.oauth.provider.type.{storage_account}.dfs.core.windows.net",
    "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider"
)
spark.conf.set(
    f"fs.azure.account.oauth2.client.id.{storage_account}.dfs.core.windows.net",
    client_id
)
spark.conf.set(
    f"fs.azure.account.oauth2.client.secret.{storage_account}.dfs.core.windows.net",
    client_secret
)
spark.conf.set(
    f"fs.azure.account.oauth2.client.endpoint.{storage_account}.dfs.core.windows.net",
    f"https://login.microsoftonline.com/{tenant_id}/oauth2/token"
)
```

```python
container_name = "bronze"
dbutils.fs.ls(f"abfss://{container_name}@{storage_account}.dfs.core.windows.net/")
```
